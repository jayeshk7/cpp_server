```python
import matplotlib.pyplot as plt

# Data
threads = [1, 2, 3, 4, 5, 6, 7, 8, 9]
throughput = [2672.05, 4576.47, 6478.78, 7623.77, 8205.35, 8350.36, 8321.16, 8378.67, 8402.7]

# Plot
plt.figure(figsize=(10, 6))
plt.plot(threads, throughput, marker='o')  # line graph with bullet points

# Labels and title
plt.xlabel("Number of Threads")
plt.ylabel("Average Throughput")
plt.title("Bottleneck Component - CPU")

# Grid (optional but looks good)
plt.grid(True)

# Show the graph
plt.show()
```

```python
import matplotlib.pyplot as plt

# Data
threads = [1, 2, 3, 4, 5, 6, 7, 8, 9]
avg_response = [0.371993, 0.434392, 0.460492, 0.522148, 
                0.606885, 0.716006, 0.830833, 0.966105, 1.060171]

# Plot
plt.figure(figsize=(10, 6))
plt.plot(threads, avg_response, marker='o')  # line graph with bullets

# Labels and Title
plt.xlabel("Number of Threads")
plt.ylabel("Average Response Time")
plt.title("Bottleneck Component - CPU")

# Optional grid for clarity
plt.grid(True)

# Display plot
plt.show()
```

```python
import matplotlib.pyplot as plt

# Actual data
threads = [2, 4, 8, 16, 32, 64, 128, 256, 512, 1024]
throughput = [899.58, 1716.69, 2376.69, 2508.96, 2559.61,
              2584.88, 2683.74, 2567.13, 2561.43, 2597.43]

# Create positions for equal spacing
positions = list(range(len(threads)))  # 0,1,2,...,9

# Plot
plt.figure(figsize=(12, 6))
plt.plot(positions, throughput, marker='o', linewidth=2)

# Label properly
plt.xlabel("Number of Threads")
plt.ylabel("Average Throughput")
plt.title("Bottleneck Component - Disk")

# Replace numeric positions with thread labels
plt.xticks(positions, threads)

# Grid
plt.grid(color='lightgray', linestyle='-', linewidth=0.5)

plt.show()
```

```python
import matplotlib.pyplot as plt

# Data
threads = [2, 4, 8, 16, 32, 64, 128, 256, 512, 1024]
avg_response = [2.25102, 2.36966, 3.43869, 6.46284, 12.6791,
                25.3621, 49.2224, 102.339, 205.291, 405.602]

# Equal spacing positions
positions = list(range(len(threads)))  # 0,1,2,...,9

# Plot
plt.figure(figsize=(12, 6))
plt.plot(positions, avg_response, marker='o', linewidth=2)

# Labels and Title
plt.xlabel("Number of Threads")
plt.ylabel("Average Response Time")
plt.title("Bottleneck Component - Disk")

# Use the thread values as x-axis labels
plt.xticks(positions, threads)

# Light grid
plt.grid(color='lightgray', linestyle='-', linewidth=0.5)

# Show graph
plt.show()
```
