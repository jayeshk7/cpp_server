```python
import matplotlib.pyplot as plt

# Data
threads = [1, 2, 3, 4, 5, 6, 7, 8, 9]
throughput = [3357.13, 6155.99, 7075.34, 8657.44, 10204.5, 10457, 10487.6, 10451.9, 10534.9]

plt.figure(figsize=(12, 7))

plt.plot(
    threads, throughput,
    color='green',
    marker='^',
    markersize=7,
    markeredgecolor='black',
    linewidth=3,
)

plt.xlabel("Number of Threads")
plt.ylabel("Avg Throughput (RPS)")
plt.title("CPU-bound Workload")

plt.grid(True, linestyle='--', alpha=0.4)

plt.tight_layout()
plt.savefig('read_all_throughput.jpg')
plt.show()
```

```python
import matplotlib.pyplot as plt

threads = [2,4,8,16,24,32,40,50,64]

# Your existing metrics
avg_response = [52.5505, 65.9624, 96.5627, 193.621, 288.964, 387.151, 483.588, 610.654, 775.584]

# New arrays
cpu_util = [31.0813, 54.587, 66.1869, 80.48, 90.803, 92.2346, 91.9416, 91.884, 91.8718]
disk_util = [0.689333, 0.582666, 0.773332, 0.599999, 0.741332, 1.356, 0.748653, 3.60247, 1.52605]

plt.figure(figsize=(12, 6))

# Plot CPU Utilization (green)
plt.plot(
    threads, cpu_util,
    color='green',
    marker='o',
    markersize=8,
    markeredgecolor='black',
    linewidth=2,
    label='CPU Utilization (%)'
)

# Plot Disk Utilization (orange)
plt.plot(
    threads, disk_util,
    color='orange',
    marker='s',
    markersize=8,
    markeredgecolor='black',
    linewidth=2,
    label='Disk Utilization (%)'
)

plt.xlabel("Number of Threads")
plt.ylabel("Utilization (%)")
plt.title("CPU & Disk Utilization vs Threads")
plt.xticks(threads)
plt.grid(color='lightgray', linestyle='-', linewidth=0.5)
plt.legend()

plt.savefig('cpu_disk_vs_threads_read_all.jpg')
plt.show()
```
